```python
# Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generating data for plots
data = np.random.normal(10, 50, 100)
data
```

```python
plt.hist(data, bins=20)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Histogram using generated data")

plt.show()
```

```python
plt.scatter(x=data, y=data)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Scatter plot using data")
plt.show()
```

```python
plt.plot(data)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("Line graph for data")
plt.show()
```
