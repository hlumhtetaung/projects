### Worldometer Data

```python
worldometer_data.info()
```

Continent missing row gives "diamond princess" region, which does not exist. Thus, missing continent will be replaced with unknown since its a ship.

```python
worldometer_data['Continent'] = worldometer_data['Continent'].fillna('Unknown')
worldometer_data['WHO Region'] = worldometer_data['WHO Region'].fillna('Unknown')
```

```python
zero_fill_cols = [
    'NewCases', 'TotalDeaths', 'NewDeaths', 'TotalRecovered',
    'NewRecovered', 'ActiveCases', 'Serious,Critical', 'TotalTests'
]
worldometer_data[zero_fill_cols] = worldometer_data[zero_fill_cols].fillna(0)

worldometer_data['Tot Cases/1M pop'] = (worldometer_data['TotalCases'] / worldometer_data['Population']) * 1_000_000
worldometer_data['Deaths/1M pop'] = (worldometer_data['TotalDeaths'] / worldometer_data['Population']) * 1_000_000
worldometer_data['Tests/1M pop'] = (worldometer_data['TotalTests'] / worldometer_data['Population']) * 1_000_000

worldometer_data.replace([np.inf, -np.inf], 0, inplace=True)
```

```python
worldometer_data.isna().sum()
```

The rest of the null values left on diamond princess are filled with .fillna(0, inplace=True), as the ship has 0 death per population.

## Data Anomalies Check
This section will check if there are odd data among the dataset, such as dates from the future or extraordinary amount of statistics.

```python
# Getting today date, date from dataframe and formatting to same format (for date checks)
today = datetime.now()
formatted_today = today.strftime("%m/%d/%Y")
```

### Covid cases by day (visualization)

```python
ccbd_visualization.info()
```

#### Date Check

```python
ccbd_visualization['Date'] = pd.to_datetime(ccbd_visualization['Date'],
                                            format='%m/%d/%Y')

# Checking date error
date_error = ccbd_visualization[ccbd_visualization['Date'] > formatted_today]
print("The number of dates that are out of range from today date:" ,
      len(date_error))
```

Non of the date exceed the current date which means the dates are correct.

#### Confirmed Check
Based on looking at the dataframe of covid cases by day, the confirmed value should be the sum of deaths, recovered, and active values.

```python
confirmed_error = ccbd_visualization[ccbd_visualization['Confirmed'] !=
                                     (ccbd_visualization['Deaths'] +
                                      ccbd_visualization['Recovered'] +
                                      ccbd_visualization['Active'])]
print("The number of confirmed deaths that doesn't match the sum of deaths, recovered, and active:", len(confirmed_error))
```
