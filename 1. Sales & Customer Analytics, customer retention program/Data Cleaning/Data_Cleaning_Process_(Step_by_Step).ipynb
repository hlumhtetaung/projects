# Data Cleaning of Retail Sales

```python
import pandas as pd
import numpy as np
from google.colab import files
```

## Checking Null Values

This block is to import and check dataset to see if there are unnecessary rows or not.

```python
df = pd.read_csv('/content/Train.csv')
```

Some null values were found in postal codes according to df.info() before the fix.

```python
df.info()
```

```python
print("Numbers of missing postal codes: ", df['Postal Code'].isna().sum())
```

Used group by fill for the missing values. Even though postal codes are missing I noticed city and other are filled out. So I used group by on city and used similar postal code from the same city.

```python
# Group by code block using city as an anchor

similar_code = df.groupby('City')['Postal Code'].first()
df['Postal Code'] = df['Postal Code'].fillna(df['City'].map(similar_code))
```

```python
print("Remaining missing postal codes:", df['Postal Code'].isna().sum())
```

Now every missing postal codes are filled with code from same city.

## Duplication Check

```python
df.duplicated().sum()
```

No duplication in dataset.

## Anomalies and Odd Data Check

```python
df.info()
```

Checking logical errors for delivery date.

```python
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%y')
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%y')

date_error = df[df['Ship Date'] < df['Order Date']]
print("Number of rows where shipping happen before order:", len(date_error))
```

Checking negative sales.

```python
print("Number of negative sale values: ", len([sale for sale in df['Sales']
                                               if sale < 0]))
```
